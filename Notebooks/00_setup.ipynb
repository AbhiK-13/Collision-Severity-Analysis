```python
# Necessary libraries: 

import pandas as pd 
from pathlib import Path
```

```python
base_path = Path(r"C:/Users/abhik/OneDrive/Desktop/WISE")
    # This base path is local to my laptop; edit as required 
    # The base path is until the WISE folder 
collision_raw_path = base_path / "Data" / "Raw" / "dft-road-casualty-statistics-collision-last-5-years.csv"
vehicle_raw_path = base_path / "Data" / "Raw" / "dft-road-casualty-statistics-vehicle-last-5-years.csv"
casualty_raw_path = base_path / "Data" / "Raw" / "dft-road-casualty-statistics-casualty-last-5-years.csv"
```

```python
print(casualty_raw_path)
print(collision_raw_path)
print(vehicle_raw_path)
```

```python
# This creates a dataframe for each of the downloaded files 

casualty_df = pd.read_csv(casualty_raw_path)
collision_df = pd.read_csv(collision_raw_path)
vehicle_df = pd.read_csv(vehicle_raw_path)
```

The warnings here indicate that the collision_df dataset has mixed types for the collision_index and collision_ref_no variables. The same applies to the generic_make_model variable in the vehicle_raw dataset. This is important to consider because it means that if we attempt merging the datasets together, observations with the same collision_index value may not be properly merged together because the values across different datasets are stored as different object types. Therefore, we should take the recommendation that has been given in the message above to force a data type option for those specific rows. 

```python
dtype_map = {'collision_index': str, 'collision_ref_no': str, 'collision_year': str, 'generic_make_model':str}

casualty_df = pd.read_csv(casualty_raw_path, dtype=dtype_map)
collision_df = pd.read_csv(collision_raw_path, dtype=dtype_map)
vehicle_df = pd.read_csv(vehicle_raw_path, dtype=dtype_map)
```

```python
# This checks that the number of different collisions that the datasets have data on are the same
# This isn't required to be the same as only the rows with matching collision reference numbers will be the same, but it is useful to see if there are any discrepancies

print("Causalty dataset:", len(pd.unique(casualty_df['collision_ref_no'])))
print("Colliison dataset:" , len(pd.unique(collision_df['collision_ref_no'])))
print("Vehicle dataset:" , len(pd.unique(vehicle_df['collision_ref_no'])))
```

```python
# This checks the shape of each dataset:

print("Collision dataset", collision_df.shape)
print("Casualty dataset", casualty_df.shape)
print("Vehicle dataset", vehicle_df.shape)

# We see that there are more observations for each unique collision (which makes sense intuitively, e.g.: for each collision there will be at least two different vehicles)
```

**Merging datasets:**

```python
# Seeing which columns are common across all 3 datasets: 
print(set(collision_df.columns) & set(vehicle_df.columns) & set(casualty_df))

# Seeing which columns are common across collision_df and vehicle_df: 
print(set(collision_df.columns) & set(vehicle_df.columns))

# Seeing which columns are common across vehicle_df and causalty_df: 
print(set(vehicle_df.columns) & set(casualty_df))

# Seeing which columns are common across collision_df and causalty_df: 
print(set(collision_df.columns) & set(casualty_df))
```

```python
# Merging the vehicle and collision datasets: 
collision_combined_1 = collision_df.merge(vehicle_df, on = ['collision_ref_no', 'collision_index', 'collision_year'])
```

```python
collision_combined_2 = collision_combined_1.merge(casualty_df, on = ['collision_ref_no', 'collision_index', 'collision_year'])
```
